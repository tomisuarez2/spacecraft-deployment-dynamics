# Derivation of the Equations of Motion for the INSAT Solar Array Deployment using Lagrange's Method and SymPy

## INSAT Solar Array Deployment

In this notebook, the equations of motion for the INSAT solar array deployment mechanism described in the following paper "Modeling and Simulation of Spacecraft Solar Array Deployment, Wie, B., Furumoto, N., Banerjee, A. K., & Barba, P. M. (1986)" are derived step by step.

The goal is to obtain a system of differential equations ready for numerical integration.

The model corresponds to the flat system shown in Fig. 7 of the article, represented as a four-bar linkage joined by hinges with two torsional springs. We will consider a case with fixed base and no slots. The spacecraft is assumed to be infinitely massive and the rotational motion of the linkage and solar panel is assumed to occur about a single axis.

Although the model used in this simulation is simple (planar, fixed based, and no slot), the simulation results provided insights into the magnitude of deployment speeds and load as a function of time and their sensitivities to various parameters of the four-bar linkage mechanism.

# 1. Imports

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp
import sympy as sm
import sympy.physics.mechanics as me

sm.init_printing()
me.init_vprinting(use_latex='mathjax')
```

# 2. System definition

The system is represented as a four-bar linkage, where only two of the bars have inertia:

- Base (Spacecraft body, has no inertia)
- Bar 1 (Yoke, has inertia)
- Bar 2 (Solar Panel Array, has inertia)
- Bar 3 (Control Rod, has no inertia)

The movement is assumed to be completely flat.

The generalized coordinates chosen are the three angles of the hinges.

- θ₁ : twist of the bar 1 respect to the base

- θ₂ : twist of the bar 2 respect to the bar 1

- θ₃ : twist of the bar 3 respect to the bar 2

Although three hinge angles are used as the generalized coordinates, they are not independent variables.

Since this model has one degree of freedom, a single coordinate, such as the angle θ₂, suffices for the description of all admissible configurations. 

```python
# Independent variable
t = me.dynamicsymbols._t

# Generalized coordinates
theta1, theta2, theta3 = me.dynamicsymbols(r'\theta_1 \theta_2 \theta_3')

## Physical parameters

# Distances
d0, d1, d2, d3 = sm.symbols('d_0 d_1 d_2 d_3', positive=True)
r1, r2 = sm.symbols('r_1 r_2', positive=True)

# Mass
m1, m2 = sm.symbols('m_1 m_2', positive=True)

# Inertia
I1, I2 = sm.symbols('I_1 I_2', positive=True)

# Hinge's stifness
k1, k2 = sm.symbols('k_1 k_2', positive=True)

# Hinge's latch up angles
theta1F, theta2F = sm.symbols('\theta_{1F} \theta_{2F}')

# Generalized coordinates vector
q = sm.Matrix([theta1, theta2, theta3])

# Generalized velocities vector
dq = q.diff(t)

# Generalized accelerations vector
ddq = dq.diff(t)
```

# 3. Reference frames

There is one referece frame asociated to each bar.

```python
# Inertial reference frame and origin (Base frame)
B0 = me.ReferenceFrame('B_0')

# Bar 1 reference frame
B1 = B0.orientnew('B_1', 'Axis', [theta1, B0.z])

# Bar 2 reference frame
B2 = B1.orientnew('B_2', 'Axis', [theta2, B0.z])

# Bar 3 reference frame
B3 = B2.orientnew('B_3', 'Axis', [theta3, B0.z])
```

# 4. Points

```python
# Inertial reference frame origin (Base - Bar 1 hinge)
H1 = me.Point(B0)
H1.set_vel(B0, 0)

# Bar 1 center of mass
G1 = H1.locatenew('G_1', r1 * B1.x)

# Bar 1 - Bar 2 hinge
H2 = H1.locatenew('H_2', d1 * B1.x)

# Bar 2 center of mass
G2 = H2.locatenew('G_2', r2 * B2.x)

# Bar 2 - Bar 3 hinge
H3 = H2.locatenew('H_3', d2 * B2.x)

# Bar 3 - Base hinge
H4 = H3.locatenew('H_4', d3 * B3.x)
```

# 5. Holonomic Constraints

We define the holonomic constraint equation in matrix form as:

$$\mathbf{\Phi}(\mathbf{q}) = \mathbf{0}$$

Where

$$\mathbf{q} = \begin{bmatrix} \theta_1 \\ \theta_2 \\ \theta_3 \end{bmatrix}$$

For this particular case, we have two constraints

$$\mathbf{\Phi} = \begin{bmatrix} \Phi_1 \\ \Phi_2 \end{bmatrix}$$

```python
hol_cons = H4.pos_from(H1) + d0 * B0.x

Phi = sm.Matrix([sm.trigsimp(hol_cons.dot(B0.x)), sm.trigsimp(hol_cons.dot(B0.y))])
Phi
```

# 6. Nonholonomic Constraints

Since it is impractical to isolate q₁ and q₃ in terms of q₂, these holonomic-constraint equations are differentiated with
respect to time and treated as nonholonomic constraints, and the dependent variables are eliminated by the method of Lagrangian multipliers.

Then

$$\frac{d\mathbf{\Phi}(\mathbf{q})}{dt} = \mathbf{A}(\mathbf{q})\dot{\mathbf{q}} = \mathbf{0}$$

```python
dPhi = Phi.diff(t)
dPhi
```

```python
A = dPhi.jacobian(dq)
A
```

# 7. Velocities

```python
G1.v2pt_theory(H1, B0, B1)

H2.v2pt_theory(H1, B0, B1)

G2.v2pt_theory(H2, B0, B2)
```

# 8. Inertia Tensors

```python
# Bar 1
Ib1 = me.inertia(B1, 0, 0, I1)

# Bar 2
Ib2 = me.inertia(B2, 0, 0, I2)
```

# 9. Rigid Bodies

```python
# Bar 1
Body1 = me.RigidBody('Bar1', G1, B1, m1, (Ib1,G1))

# Bar 2
Body2 = me.RigidBody('Bar2', G2, B2, m2, (Ib2,G2))
```

# 10. Kinetic Energy

```python
T1 = Body1.kinetic_energy(B0)

T2 = Body2.kinetic_energy(B0)

T = sm.expand(T1 + T2)

T
```

# 11. Potential Energy

```python
U = (
  + sm.Rational(1,2) * k1 * (theta1 - theta1F) ** 2
  + sm.Rational(1,2) * k2 * (theta2 - theta2F) ** 2
)

U = U.expand()
```

# 12. Lagrangian

The lagrangian of the system then is:

$$\mathcal{L = T - U}$$

```python
L = T - U
```

# 13. Euler-Lagrange Equations

As we have two constraints, we introduce two multipliers:

$$ \mathbf{\lambda} = \begin{bmatrix} \lambda_1 \\ \lambda_2 \end{bmatrix}$$

For each generalized coordinate, the corresponding Euler-Lagrange Equation is

$$ \frac{d}{dt} \left( \frac{\partial \mathcal{L}}{\partial \dot{q}_k} \right) - \frac{\partial \mathcal{L}}{\partial {q}_k} = \lambda_1 \frac{\partial\Phi_1}{\partial {q}_k} + \lambda_2 \frac{\partial\Phi_2}{\partial {q}_k} $$

where $k$ goes from 1 to 3

Thus

$$ \frac{d}{dt} \left( \frac{\partial \mathcal{L}}{\partial \dot{\mathbf{q}}} \right) - \frac{\partial \mathcal{L}}{\partial \mathbf{q}} = \mathbf{A}(\mathbf{q})^T \mathbf{\lambda} $$

A similar form of the equation is:

$$ \mathbf{M}(\mathbf{q})\ddot{\mathbf{q}} + \mathbf{h}(\mathbf{q}, \dot{\mathbf{q}}) = \mathbf{A}(\mathbf{q})^T \mathbf{\lambda} $$


```python
# Lagrange's multipliers
lam1, lam2 = sm.symbols('\lambda_1 \lambda_2')
lam = sm.Matrix([lam1, lam2])

EL = []

for qi, dqi in zip(q, dq):

    eq = (sm.diff(sm.diff(L,dqi),t) - sm.diff(L, qi))

    EL.append(sm.expand(eq))
    
EL = sm.Matrix(EL)

ddq1, ddq2, ddq3 = sm.symbols('ddq1:4')

subs_ddq = {
    ddq[0]: ddq1,
    ddq[1]: ddq2,
    ddq[2]: ddq3,
}

EL_sym = EL.xreplace(subs_ddq)

M, h = sm.linear_eq_to_matrix(EL_sym, [ddq1, ddq2, ddq3])

M = M.applyfunc(sm.cancel)
h = -h.applyfunc(sm.cancel)
```

# 14. Completing the System's Equations of Motion

The system's equations of motion are completed by adding the following equation:

$$ \mathbf{A} (\mathbf{q}) \dot{\mathbf{q}} = \mathbf{0} $$

Deriving again,

$$ \frac{d}{dt}\left[\mathbf{A}(\mathbf{q})\dot{\mathbf{q}}\right] = \mathbf{0} $$

we obtain:

$$ \mathbf{A}(\mathbf{q}) \ddot{\mathbf{q}} + \dot{\mathbf{A}}(\mathbf{q}) \mathbf{q} = \mathbf{0} $$

```python
dA = A.diff(t)
```

# 15. System's Equations of Motion in Augmented Matrix Form

Now we have

**Dynamics**
$$ \mathbf{M} \ddot{\mathbf{q}} - \mathbf{A}^T \mathbf{\lambda} = -\mathbf{h} $$

**Constraints**
$$ \mathbf{A} \ddot{\mathbf{q}} = -\dot{\mathbf{A}} \dot{\mathbf{q}} $$

Then,

$$ \boxed{ \begin{bmatrix} \mathbf{M} & - \mathbf{A}^T \\ \mathbf{A} & \mathbf{0} \end{bmatrix} \begin{bmatrix} \ddot{\mathbf{q}} \\ \mathbf{\lambda} \end{bmatrix} = \begin{bmatrix} -\mathbf{h}\\ -\dot{\mathbf{A}} \dot{\mathbf{q}} \end{bmatrix} } $$

```python
# Augmented system's matrix
KKT = M.row_join(-A.T)
KKT = KKT.col_join(A.row_join(sm.zeros(2, 2)))

# Right hand side
rhs = (-h).col_join(-dA * dq)

# Solution
solution = KKT.LUsolve(rhs)
```

# 16. Dynamic Variables Subsitution

The state of the system is the configuration space

$$ \mathbf{y} = \begin{bmatrix}\mathbf{q} \\ \dot{\mathbf{q}}\end{bmatrix}$$

```python
x1,x2,x3 = sm.symbols('x1:4')
v1,v2,v3 = sm.symbols('v1:4')

subs = {

    q[0]: x1,
    q[1]: x2,
    q[2]: x3,

    dq[0]: v1,
    dq[1]: v2,
    dq[2]: v3,

}

solution_num = solution.subs(subs)

state = sm.Matrix([x1, x2, x3, v1, v2, v3])
```

# 17. Parameters

```python
parameters = sm.Matrix([
    d0, d1, d2, d3,
    r1, r2,
    m1, m2, 
    I1, I2, 
    k1, k2, 
    theta1F, theta2F
])

# Parameter values
p_vals = np.array([
    0.15225, 1.32588, 0.08024, 1.4165, # m
    0.8915, 1.8288, # m
    8.899, 36.891, # kg
    0.9369, 43.527, # kgm^2
    0.6114, -0.776, # Nm/rad
    np.deg2rad(240), np.deg2rad(298), # rad, values after latch-up events
]) 
```

# 18. System's Function

```python
f_ddq = sm.lambdify((state, parameters), solution_num, modules="numpy", cse=True)

def dynamics(t, y, parameters):

    q1, q2, q3, dq1, dq2, dq3 = y

    ddq = f_ddq(y, parameters)

    return [dq1, dq2, dq3, ddq[0], ddq[1], ddq[2]]
```

# 19. Numerical Integration

```python
# Initial generalized coordinates and generalized velocities
q0 = [0.294, 172.817, 7.283] # Degrees
qd0 = [0,0,0] # Degrees / s

# Initial time
t0 = 0.0 # s

# Final time
tf = 10.5 # s

# Time span
t_span = [t0, tf] 

# Step
dt = 0.1

# Initial conditions.
y0 = np.deg2rad(np.array(q0 + qd0)) 

# Integrate the system
sol = solve_ivp(dynamics, t_span, y0, args=(p_vals,), method='RK45', max_step = dt, dense_output=True)

ts = sol.t
ys = sol.y
```

# 20. Plots

```python
fig, ax = plt.subplots(1, 1, sharex=True, figsize=(6, 10))

ax.plot(ts, np.rad2deg(ys[0,:]))
ax.plot(ts, np.rad2deg(ys[1,:]))
ax.plot(ts, np.rad2deg(ys[3,:]), '--')
ax.plot(ts, -np.rad2deg(ys[4,:]), '--')
ax.legend([me.vlatex(q[0], mode='inline'), me.vlatex(q[1], mode='inline'), me.vlatex(dq[0], mode='inline'), me.vlatex(-dq[1], mode='inline')])
ax.set_xlabel('Time [s]')
ax.set_ylabel('Angle [deg] or Rate [deg/s]')

fig.tight_layout()
```
